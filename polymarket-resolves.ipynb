```python
import requests
import pandas as pd

BASE_URL = "https://gamma-api.polymarket.com/markets/keyset"

def get_all_open_markets():
    all_markets = []
    cursor = None

    headers = {"Authorization": "Bearer _"}

    while True:
        params = {
            "limit": 100,
            "active": "true",
            "closed": "false",
            "include_tag": "true",
        }
        if cursor:
            params["after_cursor"] = cursor

        r = requests.get(BASE_URL, params=params, headers=headers)
        r.raise_for_status()
        data = r.json()

        markets = data.get("markets", [])
        all_markets.extend(markets)
        print(f"batch={len(markets)} | total={len(all_markets)}")

        # next_cursor is omitted on the last page
        cursor = data.get("next_cursor")
        if not cursor:
            break

    return all_markets


markets = get_all_open_markets()
```

```python
df = pd.DataFrame(markets)
df
```
